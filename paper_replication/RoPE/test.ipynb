```python
import torch
import torch.nn as nn
```

```python
def pair_rotate(x):
    x_shape = x.shape # 원래 shape 저장

    x_even = x[..., 0::2]
    x_odd = x[..., 1::2]  # 인덱스별 분리

    x_rotated = torch.stack((-x_odd, x_even), dim=-1) # 반씩 잘라서 쌓기
    return x_rotated.view(x_shape) # shape 복원

def apply_pe(x, cos, sin):
    return (x * cos) + (pair_rotate(x) * sin)

def get_rope_values(dim, seq_len, k=1, theta=10000):
    inverted_freq = k / (theta ** (torch.arange(0, dim, 2, dtype=torch.float) / dim))
    t = torch.arange(seq_len, dtype=torch.float)
    freqs = torch.outer(t, inverted_freq)
    emb = torch.cat((freqs, freqs), dim=-1)
    return freqs, emb


freqs, emb = get_rope_values(dim=4, seq_len=5, k=1, theta=10000)

print(freqs, "\n", emb)
```
