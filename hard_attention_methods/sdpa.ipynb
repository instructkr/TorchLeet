# Implement SDPA from scratch
The underlying architecture of transformer is Scaled-dot product attention, or SDPA for short.


```python
import numpy as np

def softmax(x):
    return np.exp(x)/ np.sum(np.exp(x), axis=-1, keepdims=True)

def compute_qkv(X, W_q, W_k, W_v):
    Q = X @ W_q
    K = X @ W_k
    V = X @ W_v
    return Q, K, V

def self_attention(Q, K, V):
    d_k = K.shape[1]
    attention_output = softmax((Q @ K.T)/ np.sqrt(d_k))
    attention_output = attention_output @ V
    return attention_output

```

```python
X = np.array([[1, 0], [0, 1]])
W_q = np.array([[1, 0], [0, 1]])
W_k = np.array([[1, 0], [0, 1]])
W_v = np.array([[1, 2], [3, 4]])

Q, K, V = compute_qkv(X, W_q, W_k, W_v)
output = self_attention(Q, K, V)

print(output)
```
